```python
import duckdb
import pandas as pd

ads = pd.read_csv("data/ads_master_dataset.csv")

con = duckdb.connect("ads.db")

con.execute("CREATE TABLE ads AS SELECT * FROM ads")

print("Database ready")
```

Example Analytics Queries

```python
query = """
SELECT competitor, COUNT(*) as ads_count
FROM ads
GROUP BY competitor
ORDER BY ads_count DESC
LIMIT 10
"""

print(con.execute(query).df())
```

strategy proportion

```python
query = """
SELECT strategy, COUNT(*) as count
FROM ads
GROUP BY strategy
"""
```

brand comparison

```python
query = """
SELECT competitor,
COUNT(*) * 100.0 / SUM(COUNT(*)) OVER() as percentage
FROM ads
GROUP BY competitor
"""
```
