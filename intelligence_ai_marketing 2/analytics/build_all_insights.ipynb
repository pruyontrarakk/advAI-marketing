```python
import pandas as pd
import json
import os

DATA = "/content/drive/MyDrive/Colab Notebooks/intelligence_ai_marketing/data/ads_with_strategies.csv"

OUTPUT_DIR = "/content/drive/MyDrive/Colab Notebooks/intelligence_ai_marketing/analytics/insights"

os.makedirs(OUTPUT_DIR, exist_ok=True)

df = pd.read_csv(DATA)

print("Dataset loaded:", len(df))

#################################
# 1 STRATEGY INSIGHTS
#################################

strategy_insights = {}

for strategy in df["strategy"].dropna().unique():

    subset = df[df["strategy"] == strategy]

    avg_sentiment = subset["sentiment_polarity"].mean()

    top_keywords = (
        subset["top_keywords"]
        .dropna()
        .str.split("|")
        .explode()
        .value_counts()
        .head(10)
        .index
        .tolist()
    )

    top_brands = (
        subset["competitor"]
        .value_counts()
        .head(5)
        .to_dict()
    )

    strategy_insights[strategy] = {
        "num_ads": int(len(subset)),
        "avg_sentiment": float(avg_sentiment),
        "top_keywords": top_keywords,
        "top_brands": top_brands
    }

with open(os.path.join(OUTPUT_DIR, "strategy_insights.json"), "w") as f:
    json.dump(strategy_insights, f, indent=2)

print("Strategy insights saved")

#################################
# 2 BRAND INSIGHTS
#################################

brand_insights = {}

for brand in df["competitor"].dropna().unique():

    subset = df[df["competitor"] == brand]

    strategies = (
        subset["strategy"]
        .value_counts()
        .to_dict()
    )

    avg_sentiment = subset["sentiment_polarity"].mean()

    keywords = (
        subset["top_keywords"]
        .dropna()
        .str.split("|")
        .explode()
        .value_counts()
        .head(10)
        .index
        .tolist()
    )

    brand_insights[brand] = {
        "num_ads": int(len(subset)),
        "strategies_used": strategies,
        "avg_sentiment": float(avg_sentiment),
        "top_keywords": keywords
    }

with open(os.path.join(OUTPUT_DIR, "brand_strategy_patterns.json"), "w") as f:
    json.dump(brand_insights, f, indent=2)

print("Brand insights saved")

#################################
# 3 VISUAL INSIGHTS
#################################

visual_patterns = {}

for strategy in df["strategy"].dropna().unique():

    subset = df[df["strategy"] == strategy]

    layout = subset["layout_type"].value_counts().to_dict()

    colors = (
        subset["dominant_color_1"]
        .value_counts()
        .head(5)
        .index
        .tolist()
    )

    visual_patterns[strategy] = {
        "common_layouts": layout,
        "dominant_colors": colors
    }

with open(os.path.join(OUTPUT_DIR, "visual_design_patterns.json"), "w") as f:
    json.dump(visual_patterns, f, indent=2)

print("Visual insights saved")

print("ALL ANALYTICS BUILT")
```
