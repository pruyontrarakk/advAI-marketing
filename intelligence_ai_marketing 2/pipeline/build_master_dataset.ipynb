```python
# pipeline/build_master_dataset.py

import pandas as pd


def build_master():

    print("Loading datasets...")

    cv = pd.read_csv("data/collected_ads_enriched.csv")
    nlp = pd.read_csv("data/image_text_analysis.csv")
    cluster = pd.read_csv("data/ads_with_strategies.csv")

    print("Merging CV + NLP")
    df = cv.merge(nlp, on="ad_id", how="left")

    print("Merging clustering results")
    df = df.merge(cluster[["ad_id","cluster","strategy"]], on="ad_id", how="left")

    df.to_csv("data/ads_master_dataset.csv", index=False)

    print("Master dataset created!")


if __name__ == "__main__":
    build_master()
```
